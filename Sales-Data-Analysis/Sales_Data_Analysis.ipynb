```python
import pandas as pd

df = pd.read_csv("/content/samplesuperstore.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Month"] = df["Order Date"].dt.month

monthly = df.groupby("Month")[["Sales", "Profit"]].sum()
monthly["Profit Margin"] = (monthly["Profit"] / monthly["Sales"]) * 100

monthly
```

```python
monthly["Month Name"] = pd.to_datetime(
    monthly.index, format="%m"
).month_name()

monthly
```

```python
import matplotlib.pyplot as plt
```

```python
monthly["Sales"].plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```

```python
monthly[["Sales", "Profit"]].plot(kind="bar")

plt.title("Monthly Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.show()
```

```python
insights = {
    "Highest Sales Month": monthly.loc[monthly["Sales"].idxmax(), "Month Name"],
    "Highest Profit Month": monthly.loc[monthly["Profit"].idxmax(), "Month Name"],
    "Highest Profit Margin Month": monthly.loc[monthly["Profit Margin"].idxmax(), "Month Name"],
    "Lowest Profit Margin Month": monthly.loc[monthly["Profit Margin"].idxmin(), "Month Name"]
}

insights
```

```python
print("KEY BUSINESS INSIGHTS")
print("----------------------")
print("1. November had the highest total sales.")
print("2. December had the highest total profit.")
print("3. February had the highest profit margin.")
print("4. April had the lowest profit margin.")
print("5. Technology generated the highest total profit among categories.")
print("6. Chairs had the highest total sales among sub-categories.")
print("7. Copiers generated the highest total profit among sub-categories.")
print("8. The West region had the highest total sales and profit.")
```
